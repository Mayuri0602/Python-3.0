```python
# Count of digits in a number

n = 1239
count = 0
while (n > 0):
    r = n%10
    n //= 10
    count += 1
print("Count of digits:", count)    
```

```python
# Sum of digits in a number

n = 1239
sum = 0
while( n > 0):
    r = n%10
    sum += r
    n //= 10
print("Sum of digits=", sum)    
```

```python
# Reverse of a number

n = 1239
rev = 0
while (n > 0):
    r = n%10
    rev = rev * 10 + r
    n //= 10
print("Reverse of a number:", rev)    
```

```python
# Factorial of a number

n = 6 
fact = 1
while (n > 0):
    fact = fact * n
    n = n - 1
print("Factorial of a number = ", fact)     
```

```python
n = 6
product = 1

i = 1

while i <= n:
    product = product * i
    i += 1

print("Product =", product)
```

```python
# Factors of a number

n = 12
i = 1

while (i <= n):
    if (n % i == 0):
        print(i)
    i += 1
```

```python
# Palindrome Number : 121 <-> 121, 727 <-> 727

n = 141
temp = n
rev = 0

while n > 0:
    r = n % 10
    rev = rev * 10 + r
    n = n // 10

if temp == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Fibonacci Series

n = int(input("Enter number of terms: "))

a = 0
b = 1
count = 0

while count < n:
    print(a, end=" ")
    a, b = b, a + b
    count += 1
```

```python
# Armstrong Number - 153, 370

num = int(input("Enter a number: "))

original = num
sum = 0

while num > 0:
    digit = num % 10
    sum = sum + digit ** 3
    num = num // 10

if sum == original:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```
