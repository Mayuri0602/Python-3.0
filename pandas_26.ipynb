```python
import pandas as pd

df = pd.DataFrame({
    "Category": ["A", "A", "B", "B", "C"],
    "Sales": [100, 200, 150, 250, 300],
    "Quantity": [2, 4, 3, 5, 6],
    "Profit": [20, 40, 30, 50, 60]
})


df.groupby('Category').agg({
'Sales': ['sum','mean','max'],
'Quantity': 'sum'
})
```

```python
summary = df.groupby('Category').agg(
Total_Sales=('Sales', 'sum'),
Average_Sales=('Sales', 'mean'),
Max_Sales=('Sales', 'max'),
Total_Quantity=('Quantity', 'sum')).reset_index()

summary
```

```python
# Category Wise Total Sales

pd.pivot_table(
    df,
    values='Sales',
    index='Category',
    aggfunc='sum'
)
```

```python
# City Wise Total Sales

pd.pivot_table(
    df,
    values='Sales',
    index='City',
    aggfunc='sum'
)
```

```python
# Category & City Wise Sales
pd.pivot_table(
    df,
    values='Sales',
    index='Category',
    columns='City',
    aggfunc='sum',
    fill_value=0
)
```

```python
# Category Wise Total Sales & Avg, No. of Orders
pd.pivot_table(
    df,
    values='Sales',
    index='Category',
    aggfunc=['sum', 'mean', 'count']
)
```

```python
# City & Category Avg Profit
pd.pivot_table(
    df,
    values='Profit',
    index='City',
    columns='Category',
    aggfunc='mean',
    fill_value=0
)
```

```python
# City Wise Sales for each Payment Mode
pd.pivot_table(
    df,
    values='Sales',
    index='City',
    columns='Payment_Mode',
    aggfunc='sum'
)
```

```python
# Category × Order Status
pd.pivot_table(df, values='Order_Status', 
               index='Category',
            #    columns='Order_Status', 
               aggfunc='count', 
               fill_value=0)
```
