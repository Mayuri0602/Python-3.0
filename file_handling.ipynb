```python
file = open("employees.csv", "r")

# header skip
next(file)    

# to read each line
for line in file:
    data = line.strip().split(",")

    salary = int(data[2])

    if salary > 50000:
        print(data[1], salary)

file.close()
```

```python
# Before Append
print("Before Append:")

file = open("employees.csv", "r")
print(file.read())
file.close()


# Append New Employee
file = open("employees.csv", "a")

file.write("105,Ridhi,8000\n")

file.close()


# After Append
print("\nAfter Append:")

file = open("employees.csv", "r")
print(file.read())
file.close()
```

```python
# Handle FileNotFoundError while opening a file

try:
    file = open("student.txt", "r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("File does not exist.")
```

```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))

    result = x / y

    print("Result:", result)

except ValueError:
    print("Please enter numbers only.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

```python
try:
    x = int(input("Enter first number: "))
    y = int(input("Enter second number: "))

    result = x / y

except ValueError:
    print("Invalid input.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Result:", result)

finally:
    print("Program execution completed.")
```
