```python
# Part 4: Beginner Friendly Tasks
# 1. Create a new column named total_score, which is the sum of math score,
#    reading score, and writing score.
# 2. Create a new column named average_score (total score divided by 3).
# 3. Add a column named result. A student passes if their average_score is 60 or above; otherwise, they fail.
#   (Hint: Use a conditional function or lambda expression).
# 4. Find the top 5 students based on their total_score.
# 5. Find the bottom 5 students based on their math score.
```

```python
df['total_score'] = df['math score'] + df['reading score'] + df['writing score']
df
```

```python
df['average_score'] = df['total_score'].mean()
df
```

```python
df["result"] = np.where(
    df["average_score"] >= 60,
    "Pass",
    "Fail"
)
```

```python
top_5 = df.sort_values(
    by="total_score",
    ascending=False
).head(5)

print(top_5)
```

```python
bottom_5 = df.sort_values(
    by="math score",
    ascending=True
).head(5)

print(bottom_5)
```
