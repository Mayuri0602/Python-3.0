```python
import pandas as pd

orders_data = {
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Customer_ID": ["C01", "C02", "C03", "C01", "C04",
                    "C05", "C02", "C06", "C07", "C03"],
    "Product": ["Laptop", "Mobile", "Chair", "Headphones", "Table",
                "Shoes", "Laptop", "T-Shirt", "Mobile", "Chair"],
    "Sales": [50000, 20000, 5000, 3000, 7000,
              2500, 48000, 1200, 21000, 5500]
}

orders_df = pd.DataFrame(orders_data)

print(orders_df)
```

```python
customers_data = {
    "Customer_ID": ["C01", "C02", "C03", "C04", "C05", "C08"],
    "Customer_Name": ["Amit", "Priya", "Rahul", "Neha", "Vikas", "Pooja"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "City": ["Delhi", "Mumbai", "Jaipur", "Pune", "Delhi", "Mumbai"]
}

customers_df = pd.DataFrame(customers_data)

print(customers_df)
```

```python
# Merge orders df with customers df using cust_id
merged = orders_df.merge(
    customers_df,
    on='Customer_ID'
)

print(merged)
```

```python
# Perform a left join btw orders df and customers df
merged = orders_df.merge(
    customers_df,
    on='Customer_ID',
    how='left'
)

print(merged)
```

```python
merged = orders_df.merge(
    customers_df,
    on='Customer_ID',
    how='right'
)

print(merged)
```

```python
merged = orders_df.merge(
    customers_df,
    on='Customer_ID',
    how='outer'
)

print(merged)
```

```python
# Display only order_id, cust_name, city & sales after merging

merged = orders_df.merge(
    customers_df,
    on='Customer_ID',
    how='left'
)

print(merged[['Order_ID', 'Customer_Name', 'City', 'Sales']])
```

```python
# Concat - joins two dataframes (same cols name)
import pandas as pd

df1 = pd.DataFrame({
    "Order_ID": [101, 102, 103],
    "Customer": ["Amit", "Priya", "Rahul"],
    "Sales": [5000, 7000, 6000]
})

print(df1)
```

```python
df2 = pd.DataFrame({
    "Order_ID": [104, 105, 106],
    "Customer": ["Neha", "Vikas", "Sneha"],
    "Sales": [8000, 4000, 9000]
})

print(df2)
```

```python
result = pd.concat([df1, df2])

print(result)
```

```python
result = pd.concat(
    [df1, df2],
    ignore_index=True
)

print(result)
```

```python
result = pd.concat(
    [df1, df2],
    axis=1
)

print(result)
```

```python
df3 = pd.DataFrame({
    "Order_ID": [107, 108, 109],
    "Customer": ["Rohit", "Anjali", "Karan"],
    "Sales": [4500, 7500, 6500]
})

print(df3)
```

```python
result = pd.concat(
    [df1, df2, df3],
    ignore_index=True
)

print(result)
```
